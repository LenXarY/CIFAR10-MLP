```python
import numpy as np
import joblib
from sklearn.metrics import accuracy_score
```

```python
# Path to the validation set.
TEST_DATA_PATH = "./validation_set_to_students/"

# Load test images and labels.
X_test = np.load(TEST_DATA_PATH + "images.npy")
y_test = np.load(TEST_DATA_PATH + "labels.npy")

print("Test images shape:", X_test.shape)
print("Test labels shape:", y_test.shape)
```

```python
model_files = [
    "model_seed42.joblib",
    "model_seed2026.joblib",
    "model_seed3102.joblib"
]
```

```python
accuracies = []

for model_file in model_files:
    
    # Load the trained model.
    model = joblib.load(model_file)
    
    # Predict the labels of the hidden test set.
    y_pred = model.predict(X_test)
    
    # Compute classification accuracy.
    accuracy = accuracy_score(y_test, y_pred)
    
    # Store the result.
    accuracies.append(accuracy)
    
    # Display the result.
    print(f"{model_file}: {accuracy:.4f}")
```

```python
average_accuracy = np.mean(accuracies)

print("\nIndividual accuracies:")
for model_file, accuracy in zip(model_files, accuracies):
    print(f"  {model_file}: {accuracy:.4f}")

print(f"\nAverage test accuracy: {average_accuracy:.4f}")
print(f"Average test accuracy (%): {average_accuracy * 100:.2f}%")
```
